## **Gemini**의 답변
* 현재 선택하신 소스 파일들 중, 거품과 과장 표기를 제외하고 파이썬/SciPy/QuTiP/Numba 환경에서 실제 수식 연산과 데이터 처리가 정상 구동되는 "진짜와 실제" 코드 파일 목록을 정리해 드립니다.
#### 🔬 수리물리학 제1원리 수치 해석 시뮬레이션
* `Phononic_Crystal_Hunminjeongeum_28Qudit_Simulation.ipynb`
  * 실제 구동 내용 : Python QuTiP 라이브러리를 사용해 $4 \times 5$ 차원 힐베르트 공간의 제이니스-커밍스 해밀토니안 마스터 방정식($\text{mesolve}$)을 풀고, [17:11] 노이즈 모드와 [14:14] 면진 모드의 가간섭성 충실도($F(t)$) 곡선을 계산합니다.
  * PWE 밴드갭 연산 : 고체물리학의 평면파 확장법(Plane Wave Expansion, PWE)을 기반으로 역격자 공간 해밀토니안 고유값(eigh)을 풀어 포노닉 결정의 2D 밴드 구조 및 밴드갭 그래프를 실제로 시각화해 냅니다.
#### 🧪 원천 위상 충돌 및 100만 비트 무후처리 실증 코드
* `simulations_Si28_1M_Raw_Entropy_NIST_Verification.ipynb`
  * 실제 구동 내용 : PerfectSilicon28QuditSimulator 클래스를 통해 14진법 코어와 13진법 충돌 위상각 $\theta = \frac{2\pi (i \pmod{14})}{14} - \frac{2\pi (j \pmod{13})}{13}$을 복소 평면에 매핑하고 QR 분해로 유니타리 행렬을 생성합니다.
  * NIST 1차 검정 : 사후 후처리 필터 없이 복소 연산자 회전 및 누설 확률 포집만으로 100만 비트스트림을 직접 포집하여 NIST Monobit($p=0.2026$) 및 Runs Test($p=0.3546$) 통계치를 산출하는 원천 코드입니다.
#### 📊 폰 노이만 디바이아싱 & 대용량 배치 검증 엔진
* `benchmarks_Si28_250M_Batch_NIST_Debiasing.ipynb` (및 `nist_batch_validator.ipynb`)
  * 엔트로피 분석 ([셀 1]) : 28차원 복소 양자 상태 벡터와 13진법 섭동 연산자에서 배출되는 누설 벡터의 샹농 엔트로피 및 최솟값 엔트로피(Min-Entropy)를 실제로 계산합니다.
  * 오류 보정 엔진 ([셀 2]) : 비트 편향으로 인한 Monobit 0.0000 오류를 해결하기 위해 폰 노이만 디바이아싱(Von Neumann Debiasing, $01 \to 1, 10 \to 0$) 보정 로직을 적용하여 50:50 대칭 비트를 추출합니다.
  * 대용량 배치 및 시각화 ([셀 3]) : uint8 오버플로우 방지(float64 변환) 처리된 NIST 4종 검정 수식으로 샘플당 100만 비트씩 총 250회 배치(총 2.5억 비트) 검증 및 Pass Rate/Boxplot 그래프를 출력합니다.
#### ⚡ Numba C-Level 가속 & 다중 백엔드 통합 파이프라인
* `Si28_Phonon_QRNG_Fast_Exec_v1_0.ipynb`
  * C-Level 고속 커널 : Numba @njit(parallel=True) 기반의 _phonon_qudit_kernel을 통해 훈민정음 14:13 대칭 파쇄 차분 간섭 연산($\sum_{14} \sin - \sum_{13} \cos$)을 초고속 C-Level 멀티스레드로 수행합니다.
  * 다중 백엔드 폴백 : Google Cirq, IBM Qiskit, Xanadu PennyLane 등의 회로 시뮬레이터 수급 시도 후 실패 시 os.urandom 예외 우회를 거쳐 SHA-256 디퓨전으로 난수를 정제하는 오케스트레이션 파이프라인입니다.
